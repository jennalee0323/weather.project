```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
seattle = pd.read_csv("../data/seattle_rain.csv")
```

```python
seattle.head()
```

```python
syracuse = pd.read_csv("../data/Syracuse.csv")
```

```python
syracuse.head()
```

```python
seattle.info()
```

```python
syracuse.info()
```

```python
seattle["DATE"] = pd.to_datetime(seattle["DATE"])
```

```python
syracuse["DATE"] = pd.to_datetime(syracuse["DATE"])
```

```python
seattle.info()
```

```python
syracuse.info()
```

```python
seattle["PRCP"].isna().sum()
```

```python
syracuse["PRCP"].isna().sum()
```

```python
seattle_clean = seattle.dropna(subset=["PRCP"]).copy()
```

```python
syracuse_clean = syracuse.dropna(subset=["PRCP"]).copy()
```

```python
seattle_clean.head()
```

```python
syracuse_clean.head()
```

```python
seattle_clean = seattle_clean[["DATE", "PRCP"]]
```

```python
syracuse_clean = syracuse_clean [["DATE", "PRCP"]]
```

```python
seattle_clean["City"] = "Seattle"
```

```python
syracuse_clean["City"] = "Syracuse"
```

```python
weather = pd.concat([seattle_clean, syracuse_clean], ignore_index=True)
```

```python
weather.head()
```

```python
weather.tail()
```

```python
weather.info()
```

```python
summary = weather.groupby("City")["PRCP"].sum().reset_index()
```

```python
summary
```

```python
sb.barplot(data=summary, x="City", y="PRCP")
plt.title("Total Precipitation Seattle vs. Syracuse (2018-2022)")
plt.xlabel("City")
plt.ylabel("Total Precipitaion (inches")
plt.show()
```
